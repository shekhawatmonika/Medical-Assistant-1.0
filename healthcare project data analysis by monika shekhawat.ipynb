# Descriptive Statistics and Aggregation

```python
#6.	How do you calculate the percentage of patients who did not show up for their appointment?
no_show_percentage = (df['No-show'] == 'Yes').mean() * 100
print(f"No-show percentage: {no_show_percentage:.2f}%")
```

```python
#7.	How can you group the data by gender and count the no-shows in each group?
no_show_by_gender = df[df['No-show'] == 'Yes'].groupby('Gender').size()
print(no_show_by_gender)
```

```python
#8.	How would you find the average age of patients who showed up vs. those who didn’t?
avg_age_show = df[df['No-show'] == 'No']['Age'].mean()
avg_age_no_show = df[df['No-show'] == 'Yes']['Age'].mean()
print(f"Average age of patients who showed up: {avg_age_show:.2f}")
print(f"Average age of patients who didn’t show up: {avg_age_no_show:.2f}")
```

```python
#9.	How can you count how many patients received SMS reminders but still didn’t show up?
sms_no_show = df[(df['SMS_received'] == 1) & (df['No-show'] == 'Yes')].shape[0]
print(f"Patients who received SMS reminders but still didn’t show up: {sms_no_show}")
```

```python
#10.	How would you determine the neighborhood with the highest no-show rate?
no_show_rate_by_neighborhood = df.groupby('Neighbourhood')['No-show'].apply(lambda x: (x == 'Yes').mean())
print(no_show_rate_by_neighborhood.sort_values(ascending=False).head(1))
```

# Date and Time Analysis

```python
#11.	How would you convert ScheduledDay and AppointmentDay to datetime objects?
df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay'])
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay'])
df
```

```python
#12.	How do you calculate the number of days between the scheduling date and appointment date?

df['days_between'] = (df['AppointmentDay'] - df['ScheduledDay']).dt.days
df
```

```python
#13.	How would you find the average scheduling lead time for no-shows vs. shows?
avg_lead_time_show = df[df['No-show'] == 'No']['days_between'].mean()
avg_lead_time_no_show = df[df['No-show'] == 'Yes']['days_between'].mean()
print(f"Average lead time for shows: {avg_lead_time_show:.2f} days")
print(f"Average lead time for no-shows: {avg_lead_time_no_show:.2f} days")
```

```python
#14.	How can you filter appointments that were scheduled on the same day as the appointment?

same_day_appointments = df[df['ScheduledDay'].dt.normalize() == df['AppointmentDay'].dt.normalize()]
print(same_day_appointments)
```

# Data Cleaning and Preparation


```python
#15.	How do you handle incorrect or outlier values in the Age column (e.g., negative values)?
incorrect_ages = df[df['Age'] <0]
print(incorrect_ages)
```

```python
incorrect_ages = df[df['Age'] < 0]
df = df.drop(incorrect_ages.index)
print("Cleaned DataFrame (no negative ages):")
print(df[df['Age'] < 0]) 
```

```python
#16.	How would you rename columns to follow snake_case naming conventions (e.g., No-show to no_show)?
df.rename(columns={'No-show': 'no_show'},inplace=True)
df
```

```python
#17.	How would you convert the No-show column into a Boolean column?
#df['no_show'] = df['no_show'].map({'Yes': True, 'No': False}) # this is not working
df['no_show'] = df['no_show'].apply(lambda x: True if x == 'Yes' else False)
df
```

```python
#18.	How can you drop columns that are not relevant to the analysis (e.g., AppointmentID)?
df = df.drop('AppointmentID', axis=1)
df
```
